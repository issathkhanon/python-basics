# Setup

Visual Studio Code, Python, Cursor, and Git, before the first Python lesson.

Cursor and Visual Studio Code have the same structure. Cursor is an editor built on top of Visual Studio Code.

This is the setup lesson. Start with the three getting-started steps: install Visual Studio Code, install Python, then make a virtual environment and run a hello-world file. The later sections are the terminal commands from the setup notes. Run those in a terminal. A notebook cell is the wrong place for `git config` and `brew install`.


## Getting started: Visual Studio Code and Python

Three steps. Install the editor, install Python, then run one file inside a virtual environment.

### 1. Download Visual Studio Code and install it

Open [https://code.visualstudio.com](https://code.visualstudio.com) and download the installer for your computer.

On Windows, open the downloaded file. Click **Next**, **Next**, then **Install**, then **Finish**. On the options page, leave **Add to PATH** checked so the terminal can run `code`.

On a Mac, open the downloaded `.dmg` and drag **Visual Studio Code** into **Applications**.

Open Visual Studio Code. The window is the editor. **File → Open Folder** opens a project folder. **Terminal → New Terminal** opens a terminal at the bottom of that same window.

### 2. Install Python

Open [https://www.python.org/downloads/](https://www.python.org/downloads/) and download Python 3.12.

On Windows, open the installer. Check **Add python.exe to PATH**, then click **Install Now**. When it finishes, click **Close**.

On a Mac, open the installer package and click **Continue**, then **Install**.

Close the terminal if it was already open, then open a new one with **Terminal → New Terminal**. Check the install:

```bash
python --version
```

A sample result is `Python 3.12.16`. On a Mac the command is often `python3 --version`.

### 3. Create a virtual environment and run hello world

Python from step 2 has to be installed first. `python -m venv` and `python hello.py` both use that install. Without it, the terminal says `python` is not found.

In Visual Studio Code, use **File → Open Folder** and choose a new empty folder, for example `hello`.

Open **Terminal → New Terminal** and create the environment:

```bash
python -m venv .venv
```

On a Mac, if `python` is not found, use `python3 -m venv .venv`.

`.venv` is a folder with its own Python. Turn it on:

```bash
source .venv/bin/activate
```

On Windows Command Prompt the command is `.venv\Scripts\activate`. The prompt then starts with `(.venv)`.

Click the Python version in the status bar, or use **Python: Select Interpreter** from the Command Palette, and choose the interpreter inside `.venv`.

Create a file named `hello.py` and put this line in it:

```python
print("Hello, world")
```

Run it from the same terminal:

```bash
python hello.py
```

The terminal prints `Hello, world`.


## 1. Open this folder in Cursor

`PATH` is the list of folders where the terminal looks for a command. Cursor's terminal command lives inside the Cursor app. These two lines add that folder to `PATH` and reload the file:

```bash
echo 'export PATH="$PATH:/Applications/Cursor.app/Contents/Resources/app/bin"' >> ~/.bash_profile
source ~/.bash_profile
```

`>>` adds the line to the end of `~/.bash_profile`. `source` reads that file in the current terminal.

```bash
cursor .
```

`cursor .` opens the current folder in Cursor. The dot means this folder. Visual Studio Code uses the same idea with `code .` after the `code` command is on `PATH`.


## 2. Install Python 3.12 and make a private environment

```bash
brew install python@3.12
python3.12 --version
```

Homebrew installs Python 3.12. The version command uses two hyphens. A sample result is `Python 3.12.16`.

```bash
python3.12 -m venv .venv312
source .venv312/bin/activate
```

`venv` creates a folder named `.venv312` with its own Python and its own `pip`. `source` turns that environment on. The prompt then starts with `(.venv312)`.

```bash
which python
python --version
```

`which python` prints the Python that will run. After activation it points inside `.venv312`. `python --version` then prints `Python 3.12.16`, because `python` means the environment's Python.

A virtual environment is a private Python for one project. Two projects can install different versions of the same library and neither one breaks the other.

```bash
pip install -r requirements.txt
pip list
pip freeze > requirements.txt
```

`requirements.txt` is the project's package list. One name per line, for example `flask` and `pytest`. Later a line can pin a version, such as `flask==3.1.3`.

`pip install -r requirements.txt` installs every line. `pip list` shows what is installed. `pip freeze > requirements.txt` writes the exact installed versions back into the file. `>` replaces the file.

```bash
deactivate
```

`deactivate` turns the environment off. The prompt loses `(.venv312)`.

A first program can be one line, `print(1 + 1)`, saved as `hello.py`. From the folder that contains it:

```bash
python hello.py
```

The terminal prints `2`.


## 3. Git: your name, then a repo

A repo is a folder Git tracks. The history lives in a hidden `.git` folder. The path of one change is: working folder, `git add`, staging area, `git commit`, local repo, `git push`, GitHub. GitHub comes back with `git clone` the first time and `git pull` after that. `git fetch` downloads commits and leaves your files as they are. `git pull` is fetch plus merge.

Set the name and email used on every repo. This course uses Issath Khan.

```bash
git config --global user.name "Issath Khan"
git config --global user.email "issath@example.com"
git config --global init.defaultBranch main
git config --global --list
git config --global --edit
```

`user.name` and `user.email` are stored for every repo. `init.defaultBranch main` makes a new repo start on `main`. `--list` prints the settings. `--edit` opens `~/.gitconfig`.

```bash
git config user.name
git config --show-origin --get user.email
git config --list --show-origin
```

Without `--global`, `git config user.name` shows the name this repo will use. `--show-origin` adds the file the value came from.

Repos inside `~/Desktop/lab/` can use a second identity. In `~/.gitconfig`, add:

```text
[includeIf "gitdir:~/Desktop/lab/"]
    path = ~/.gitconfig-personal
```

Create `~/.gitconfig-personal` in the home folder, not inside `lab`:

```bash
vi ~/.gitconfig-personal
cat ~/.gitconfig-personal
```

The file contains:

```text
[user]
    name = Issath Khan
    email = issath@example.com
```

A repo under `~/Desktop/lab/` then uses that name and email. A local setting inside one repo's `.git/config` wins over these files.


## 4. The first commit and the first push

Work inside the project folder, not the parent `lab` folder.

```bash
cd pythonbasics
git init
```

`git init` creates `.git`. A sample line is `Initialized empty Git repository in .../pythonbasics/.git/`.

```bash
git config --show-origin --get user.name
```

A sample answer is `file:/Users/<me>/.gitconfig-personal    Issath Khan`.

```bash
echo "# Python basics" > README.md
git status
```

`>` writes the heading into `README.md` and replaces the file. `git status` then says `On branch main`, `No commits yet`, and lists `README.md` as untracked.

```bash
git add README.md
git commit -m "Add the project readme"
```

`git add` stages that file. `git add .` stages every change in the folder. `git commit -m` saves the staged snapshot. A sample result is `[main (root-commit) 9b693e4] Add the project readme`.

```bash
git remote add origin https://github.com/issathkhanon/pythonbasics.git
git remote -v
git push -u origin main
```

`origin` is the nickname for the GitHub URL. `git remote -v` prints the fetch and push URLs. `git push -u origin main` sends `main` and links it, so later `git push` and `git pull` know the remote branch. GitHub over HTTPS asks for the username `issathkhanon` and a personal access token as the password, not the account password. A sample ending is `branch 'main' set up to track 'origin/main'`.


## 5. A second commit, a clone, and the diff

```bash
vi README.md
vi notes.txt
git status
git restore README.md
```

`git status` shows `README.md` as modified and `notes.txt` as untracked. `git restore README.md` throws away the unstaged edit. There is no undo for that restore. Edit the file again, then:

```bash
git add .
git commit -m "Add the class notes"
git push
```

The link from `-u` is already set, so plain `git push` is enough. A sample range is `9b693e4..b0ca329  main -> main`.

A second copy of the repo:

```bash
git clone https://github.com/issathkhanon/pythonbasics.git
cd pythonbasics
vi scores.txt
git add scores.txt
git commit -m "Add the score file"
git push
```

`git clone` copies the history into a new folder and links `main`. A sample commit line is `[main 00bed67] Add the score file`.

Unstage, then look at the staged diff:

```bash
echo "Issath is on the register." > register.txt
git add .
git restore --staged register.txt
git status
git add .
git reset register.txt
git add .
git diff
git diff --staged
git commit -m "Add the class register"
git status
git push
```

`git restore --staged` and `git reset <file>` both take the file out of the staging area and leave the file on disk. After `git add`, plain `git diff` is empty because it shows only unstaged edits. `git diff --staged` shows the staged new file:

```text
diff --git a/register.txt b/register.txt
new file mode 100644
--- /dev/null
+++ b/register.txt
@@ -0,0 +1 @@
+Issath is on the register.
```

`git status` then says `Your branch is ahead of 'origin/main' by 1 commit`. `git push` publishes it. A sample range is `00bed67..d0f8712  main -> main`.

```bash
git rm --cached <file>
```

`git rm --cached` unstages a new file before its first commit.

```bash
git log
git log --oneline
git log --oneline --graph --all
```

`git log` prints the hash, author, date, and message. `--oneline` is one line per commit. `--graph --all` draws every branch.


## 6. A branch, a fast-forward, and two developers

```bash
git branch
git branch -a
git branch issath
git switch issath
git switch -c lessons
git branch -M main
```

`git branch` lists local branches. The star marks the current one. `git branch -a` also lists `remotes/origin/...`. `git branch issath` creates a branch and stays where you are. `git switch issath` moves to it. `git switch -c lessons` creates the branch and moves to it. `git branch -M main` renames the current branch to `main`.

```bash
echo "Lesson list" > lessons.txt
git add lessons.txt
git commit -m "Add the lesson list"
git push -u origin lessons
git switch main
git pull
git merge lessons
git push
git log --oneline
```

The push prints a pull-request link, `https://github.com/issathkhanon/pythonbasics/pull/new/lessons`, and `branch 'lessons' set up to track 'origin/lessons'`. On a team, open that pull request and merge it on GitHub, then `git pull` on `main`.

The local sample merge is a fast-forward. `main` had no newer commit, so Git moved `main` forward to `70350a0` and did not make a merge commit:

```text
Updating d0f8712..70350a0
Fast-forward
lessons.txt | 1 +
```

`git log --oneline` then shows:

```text
70350a0 (HEAD -> main, origin/main, origin/lessons, origin/HEAD, lessons) Add the lesson list
d0f8712 Add the class register
00bed67 Add the score file
b0ca329 Add the class notes
9b693e4 Add the project readme
```

```bash
git branch issath
git branch khan
git push -u origin issath khan
```

One push sends both branches and links each one. The remote lines are `issath -> issath` and `khan -> khan`.

```bash
git branch -d issath
```

`git branch -d` deletes a local branch that is already merged.

In the first copy of the repo, bring in the commits the second copy pushed:

```bash
cd pythonbasics
git pull
git branch -a
```

`git pull` works because the first push used `-u`. Without that link, Git says there is no tracking information. Set it once:

```bash
git branch --set-upstream-to=origin/main main
git pull origin main
```

`git branch -a` then lists `main` and the remote branches `origin/issath`, `origin/khan`, `origin/lessons`, and `origin/main`.


## 7. Two people edit the same lines

Khan edits the readme:

```bash
git switch khan
vi README.md
git add README.md
git commit -m "Update the readme from Khan"
git push
git switch main
git pull
git merge khan
git push
```

`git switch khan` creates the local branch and links it: `branch 'khan' set up to track 'origin/khan'`. The merge into `main` is a fast-forward, `70350a0..7edbd4b`.

Issath changes the same heading:

```bash
git switch issath
vi README.md
git add README.md
git commit -m "Update the readme from Issath"
git push
git switch main
git pull
git merge issath
```

`git merge issath` stops:

```text
CONFLICT (content): Merge conflict in README.md
Automatic merge failed; fix conflicts and then commit the result.
```

`cat README.md` shows both versions:

```text
<<<<<<< HEAD
# Python basics, Khan
=======
# Python basics, Issath
>>>>>>> issath
```

Keep the heading you want and delete the three marker lines. Then:

```bash
grep -c '<<<<<<<' README.md
git add README.md
git commit -m "Merge the readme headings"
git push
```

`grep -c` counts leftover markers. `0` means the file is clean. `git add` marks the conflict resolved. The commit is the merge commit, for example `0ec905c`. Do not run `git merge` a second time.

```bash
git merge --abort
```

`git merge --abort` returns the repo to the moment before the merge.


## 8. Small shell commands from the same session

```bash
echo "text" > file
echo "text" >> file
cat file
pwd
ls -a
grep -c '<<<<<<<' file
vi file
nano file
```

`>` replaces the file. `>>` adds to the end. `cat` prints the file. `pwd` prints the current folder. `ls -a` lists hidden names, including `.git`. `grep -c` counts matching lines. `vi` and `nano` open a file in the terminal.
