## 1. Names are case-sensitive

Python treats capital and lowercase letters as different characters. `name` and `Name` are two variables. Assigning one leaves the other unchanged.


````python
name = "Issath"
Name = "Khan"
print(name)
print(Name)
````

The first `print` shows the value stored in `name`. The second shows the value stored in `Name`. Both exist at the same time.


## 2. Indentation marks a block

Python uses indentation to decide which lines belong together. A line indented under `if` runs only when the condition is true. A line moved back to the left runs afterward, whether the condition was true or false.

Use four spaces for each level. Keep that spacing consistent in a file.


````python
x = 5
if x > 2:
    print("x is greater than 2")
print(x)
````

`x > 2` is true, so the indented `print` runs. `print(x)` sits at the left edge, so it runs next and shows `5`.


## 3. A backslash continues a line

A `\` as the last character on a line tells Python that the statement continues on the next line. Nothing, including a space, can follow the backslash on that line.

Parentheses are the form you will use most often. A statement inside `( )` can break across lines without a backslash.


````python
y = 1 + 2 + 3 \
+ 4 + 5
print(y)

total = (1 + 2 + 3
         + 4 + 5)
print(total)
````

Both expressions add the same five numbers, so each `print` shows `15`.


## 4. Several statements on one line

A semicolon separates statements written on the same line. Python runs them from left to right.

In everyday code, put each statement on its own line. A separate line is easier to read, edit, and find when something goes wrong. The semicolon form is here so you can recognize it.


````python
x = 5; y = 10; z = x + y; print(z)
````

Python stores `5` in `x`, stores `10` in `y`, adds them into `z`, then prints `15`.


## 5. A variable is a name for a value

`=` assigns the value on the right to the name on the left.

- `age = 30` stores an integer. An integer is a whole number.
- `name = "Issath"` stores a string. A string is text, and text is written in quotes.

The name is a label. The value is what Python keeps.


````python
age = 30
name = "Issath"
print(age)
print(name)
````

## 6. Ask Python for the type

`type(...)` reports the kind of value a name currently holds.

When the last line of a cell is a value, Jupyter shows that value under the cell. You can read the type there without calling `print`. In this notebook, an integer is shown as `int` and text is shown as `str`.


````python
type(age)
````

````python
type(name)
````

## 7. The same name can hold a new kind of value

Python decides the type from the value you assign. Assign a new value, and the name points at that new value. The previous value is left behind.

Each assignment below replaces what `a` means:

| Value | Type | What it is |
| --- | --- | --- |
| `10.5` | float | A number with a decimal point |
| `"khan"` | str | Text |
| `True` | bool | A boolean, either `True` or `False` |
| `None` | NoneType | The absence of a value |
| `[1, 2, 3, 4, 5]` | list | An ordered collection in square brackets |
| `(1, 2, 3, 4, 5)` | tuple | An ordered collection in parentheses |


````python
a = 10.5
print(a)
a = "khan"
print(a)
a = True
print(a)
a = None
print(a)
a = [1, 2, 3, 4, 5]
print(a)
a = (1, 2, 3, 4, 5)
print(a)
````

Each `print` shows the value from the assignment just above it. The name `a` is reused on purpose so you can see the type change.


## 8. A block can sit inside another block

An `if` inside another `if` is a nested block. A line runs when every condition that contains it is true. Read the indentation from the outside inward:

```python
if True:                          # enter the outer block
    print("inside true block")
    if False:                     # skip the line indented under this if
        print("inside false block")
    print("outside false block")  # still inside the outer block
print("outside the true block")   # outside both blocks
```


````python
if True:
    print("inside true block")
    if False:
        print("inside false block")
    print("outside false block")
print("outside the true block")
````

`inside false block` is missing from the output because that line is indented under `if False`. The other three prints run.
