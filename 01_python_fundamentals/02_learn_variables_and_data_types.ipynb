## 1. Store each value under a name that describes it

A variable name can use letters, digits, and underscores. Start it with a letter. `age`, `name`, and `is_student` say what the value means, which makes the later lines easier to read.

- `30` is an integer
- `"Issath"` is a string
- `True` is a boolean


````python
age = 30
name = "Issath"
is_student = True
print(age)
print(name)
print(is_student)
````

## 2. A type hint documents your intention

`age: int = 30` tells a reader, and tools such as an editor, that `age` is meant to hold an integer. Python still stores the value and runs the program. The hint is a note. It is there for people.


````python
age: int = 30
print(age)
````

## 3. Convert a value from one type to another

`int()`, `float()`, `str()`, and `bool()` each build a new value of that type. The lines below reuse `a` so you can follow one value as it changes.

Read the output from top to bottom:

1. `25.5` is a float.
2. `int(25.5)` keeps the whole number `25` and drops the decimal part.
3. `float(25)` builds `25.0` again. `print(type(...))` shows the type as `<class 'float'>`.
4. `str(25)` builds the text `"25"`.
5. `bool("25")` is `True`. Text converts to `True` when it contains at least one character. Empty text, `""`, converts to `False`.
6. `None` has the type `NoneType`. It means no value.
7. A number written with a decimal point, such as `1.5`, has the type `float`.


````python
a = 25.5
print(a, type(a))
a = int(a)
print(a, type(a))
print(type(float(a)))
a = str(a)
print(a, type(a))
a = bool(a)
print(a, type(a))
print(type(None))
print(type(1.5))
````

## 4. `input()` always returns text

`input("Enter Your Age:")` pauses the cell, shows the message, and waits for the keyboard. Whatever you type comes back as a string. Digits typed on the keyboard are still text until you convert them.

Run the cell, type an age, and press **Enter**. A sample run where the person typed `15` looks like this:

```text
15 <class 'str'>
```


````python
age = input("Enter Your Age:")
print(age, type(age))
````

## 5. Convert input before you use it as a number

Wrap `input()` in `int()` when you need a whole number. The cell still waits for you to type. A sample run where the person typed `24` looks like this:

```text
24 <class 'int'>
```

`int()` accepts the text of a whole number, such as `"24"`. `"24.5"` and `"Khan"` raise `ValueError`, because `int()` needs digits with no decimal point and no letters.


````python
age = int(input("Enter Your Age:"))
print(age, type(age))
````

## 6. A small calculator

This cell reads two whole numbers and prints four results:

- `total` uses `+` to add
- `difference` uses `-` to subtract
- `product` uses `*` to multiply
- `quotient` uses `//` to divide and keep the whole number

`//` is floor division. `20 // 10` is `2`. The next lesson shows `/`, which keeps the decimal.

A sample run with `20` and then `10`:

```text
30
10
200
2
```


````python
num1 = int(input("Enter First Number:"))
num2 = int(input("Enter Second Number:"))
total = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 // num2
print(total)
print(difference)
print(product)
print(quotient)
````
