## 1. `+` needs two values of the same kind

`+` between strings joins text. `+` between numbers adds them. `"Hello"` is text and `5` is an integer, so this cell raises `TypeError`.

Run it and read the last line of the error. That line names the two types Python refused to combine.


```python
result = "Hello" + 5
print(result)
```

## 2. Convert the number, then join the text

`str(5)` builds the text `"5"`. `"Hello" + "5"` joins those two strings into `"Hello5"`.


```python
result = "Hello" + str(5)
print(result)
```

## 3. Arithmetic operators

With `a = 4` and `b = 2`:

| Expression | Name | Result | Meaning |
| --- | --- | --- | --- |
| `a + b` | addition | `6` | 4 plus 2 |
| `a - b` | subtraction | `2` | 4 minus 2 |
| `a * b` | multiplication | `8` | 4 times 2 |
| `a / b` | division | `2.0` | 4 divided by 2, kept as a float |
| `a % b` | remainder | `0` | leftover after dividing 4 by 2 |
| `a ** b` | power | `16` | 4 to the power of 2 |
| `a // b` | floor division | `2` | 4 divided by 2, kept as a whole number |

`/` always produces a float, which is why the result is `2.0` even though 4 divides evenly by 2.


```python
a = 4
b = 2
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
print(a ** b)
print(a // b)
```

## 4. Comparison operators

A comparison asks a question and answers with `True` or `False`.

| Expression | Question | Result here |
| --- | --- | --- |
| `a == b` | Are the values equal? | `False` |
| `a != b` | Are the values different? | `True` |
| `a > b` | Is `a` greater? | `True` |
| `a < b` | Is `a` smaller? | `False` |

`==` compares values. `=` assigns a value. They do different jobs.


```python
a = 4
b = 2
print(a == b)
print(a != b)
print(a > b)
print(a < b)
```

## 5. Strings compare one character at a time

Python compares text from the first character onward.

- `"khan" == "khan"` is `True` because every character matches.
- `"khan" > "khan"` and `"khan" < "khan"` are `False` because the strings are equal.
- `>=` and `<=` are `True` for equal strings, because "greater or equal" and "smaller or equal" include equality.

`"Khan" > "khan"` is `False`. Each character has a Unicode number, and an uppercase letter has a smaller number than the same lowercase letter. `K` comes before `k`, so `"Khan"` is the smaller string.


```python
print("khan" > "khan")
print("khan" < "khan")
print("khan" >= "khan")
print("khan" <= "khan")
print("khan" == "khan")
print("khan" != "khan")
print("Khan" > "khan")
```

```python
print(ord("K"))
print(ord("k"))
```

`ord()` returns that Unicode number. `K` is `75` and `k` is `107`, so `K` is smaller than `k`. That is why `"Khan" > "khan"` printed `False`.


## 6. `/` keeps the fraction, `//` keeps the whole number

`21 / 5` is `4.2`. The decimal stays, and the result is a float.

`21 // 5` is `4`. Floor division drops the fraction and keeps how many whole times 5 fits into 21.


```python
21 / 5
```

```python
21 // 5
```

## 7. Logical operators

`and`, `or`, and `not` combine booleans.

With `a = True` and `b = False`:

| Expression | Result | Why |
| --- | --- | --- |
| `a and b` | `False` | `and` is `True` only when both sides are `True` |
| `a or b` | `True` | `or` is `True` when at least one side is `True` |
| `not a` | `False` | `not` flips the value |
| `a and not b` | `True` | `not b` is `True`, and `True and True` is `True` |


```python
a = True
b = False
print(a and b)
print(a or b)
print(not a)
print(a and not b)
```
