## 1. `if`, `elif`, and `else` choose one path

Python checks the conditions from top to bottom and runs the first true branch. The other branches are skipped.

- `if` is the first question.
- `elif` is the next question, checked only when every question above it was false.
- `else` runs when every question was false. It has no condition of its own.

With `age = 18`, the first condition `age >= 18` is true, so this cell prints the adult message.


```python
age = 18
if age >= 18:
    print("You are an adult")
elif age >= 13:
    print("You are a teenager")
else:
    print("You are a child")
```

Change the stored age and the chosen branch changes with it. `15` fails `age >= 18` and passes `age >= 13`, so the `elif` branch runs.


```python
age = 15
if age >= 18:
    print("You are an adult")
elif age >= 13:
    print("You are a teenager")
else:
    print("You are a child")
```

`10` fails both comparisons. The `else` branch runs.


```python
age = 10
if age >= 18:
    print("You are an adult")
elif age >= 13:
    print("You are a teenager")
else:
    print("You are a child")
```

## 2. `while` repeats while a condition is true

A `while` loop checks its condition before every pass. When the condition is true, the indented block runs. Then Python checks again.

This loop starts `count` at `1` and continues while `count` is less than `10`, so it prints `1` through `9`.

After each number, `%` gives the remainder. A remainder of `0` means the number is even. Any other remainder means it is odd. `count += 1` stores the next number. That line is what lets the condition become false and the loop end.


```python
count = 1
while count < 10:
    print(count)
    if count % 2 == 0:
        print("Even")
    else:
        print("Odd")
    count += 1
```

Each number is printed, then its label. `1` is odd, `2` is even, and the pairs continue through `9`.

The increase sits at the end of the block, so the label describes the number just printed. On the next pass, `count` is already one larger.


## 3. `for` visits each value in a sequence

`range(1, 4)` produces `1`, then `2`, then `3`. The stop value `4` is left out. `for` assigns each of those values to `i` and runs the block once per value.


```python
for i in range(1, 4):
    print(i)
```

## 4. `break` leaves the loop immediately

`break` stops the loop that contains it. The lines after `break` in that loop do not run, and there is no next pass.

Here the loop is ready to visit `1` through `5`. When `number` is `4`, `break` ends the loop, so `4` and `5` are never printed.


```python
for number in range(1, 6):
    if number == 4:
        break
    print(number)
```

## 5. A loop can sit inside another loop

The inner `for` runs its whole sequence once for each value of the outer `for`.

`else` on a `for` runs when that loop finishes without `break`. A `break` skips that `else`.

Read this cell as a search for prime numbers from `1` up to, but not including, `10`. A prime number has no divisor other than `1` and itself.

- `i` takes each value from `range(1, 10)`: `1` through `9`.
- The inner loop runs only when `i > 1`, so `1` is skipped.
- `j` starts at `2` and stops before `i`, so it tries possible divisors of `i`.
- If `i % j == 0`, then `j` divides `i` evenly and `break` leaves the inner loop. That `i` is not printed.
- If no divisor causes a `break`, the inner `else` prints `i`.


```python
for i in range(1, 10):
    if i > 1:
        for j in range(2, i):
            if i % j == 0:
                break
        else:
            print(i)
```

The cell prints `2`, `3`, `5`, and `7`.

- `1` is skipped because `i > 1` is false.
- `2` has an empty inner range, `range(2, 2)`. The inner loop never hits `break`, so `else` prints `2`.
- `3` tries `j = 2`. `3 % 2` is `1`, so there is no `break`, and `else` prints `3`.
- `4` tries `j = 2`. `4 % 2` is `0`, so `break` runs and `4` is not printed.
- `5` is not divisible by `2`, `3`, or `4`, so `else` prints `5`.
- `6` is divisible by `2`, so `break` skips the print.
- `7` has no divisor from `2` through `6`, so `else` prints `7`.
- `8` is divisible by `2`, and `9` is divisible by `3`. Each one hits `break` before `else`.


## Print prime numbers with a `for` loop

This cell runs that same search on its own. `range(1, 10)` supplies `1` through `9`. The numbers that finish the inner loop without `break` are the primes in that range, so the cell prints `2`, `3`, `5`, and `7`.


```python
for i in range(1, 10):
    if i > 1:
        for j in range(2, i):
            if i % j == 0:
                break
        else:
            print(i)
```
