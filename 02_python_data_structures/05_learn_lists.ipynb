# Lists

An ordered collection you can change.

This is lesson 5. A list keeps items in order, and you can replace, add, or remove those items later. Square brackets `[ ]` create a list.

Run each code cell with **Shift+Enter**.

In this lesson you will:

- Create an empty list and read items by position
- Take a slice, including a slice that steps through the list
- Replace, insert, remove, sort, and reverse items
- Visit every item with `for`
- Build a new list with a loop and with a comprehension

## Picture of the idea

A list keeps values in order. The first position is index 0. A later line can replace the value in a position.

![A list is ordered positions starting at index 0](diagrams/list_index.png)


## 1. An empty list

`[]` creates a list with no items. `type` reports `list`.


```python
fruits = []
print(fruits)
print(type(fruits))
```

## 2. Positions and slices

This list has four items. Counting starts at `0`.

| Position | Item |
| --- | --- |
| `0` or `-4` | `apple` |
| `1` or `-3` | `banana` |
| `2` or `-2` | `cherry` |
| `3` or `-1` | `grapes` |

A slice `start:stop` includes `start` and stops before `stop`. Leave out `start` to begin at the first item. Leave out `stop` to continue through the last item. A third number is the step: `::2` keeps every second item, and `::-1` walks backward.


```python
fruits = ["apple", "banana", "cherry", "grapes"]
print(fruits)
print(fruits[1])
print(fruits[-1])
print(fruits[1:3])
print(fruits[:2])
print(fruits[2:])
print(fruits[-3:])
print(fruits[-3:-1])
print(fruits[-1::-1])
print(fruits[::2])
print(fruits[::-1])
print(fruits[1:4:2])
print(fruits[::])
print(fruits[:])
```

Read the prints in order:

- The whole list, then `banana` at index `1`, then `grapes` at the last index.
- `1:3` is `banana` and `cherry`. The stop index `3` is left out.
- `:2` is the first two items. `2:` is everything from `cherry` onward.
- `-3:` starts at `banana` and continues to the end. `-3:-1` stops before the last item, so `grapes` is left out.
- `-1::-1` and `::-1` both walk from the end back to the start.
- `::2` keeps `apple` and `cherry`. `1:4:2` starts at `banana` and steps by 2, so it keeps `banana` and `grapes`.
- `::` and `:` both mean the whole list.


## 3. Change the list

A list is mutable. These lines change `fruits` itself:

- `fruits[1] = "mango"` replaces one item.
- `fruits[1:3] = ["orange", "pineapple"]` replaces two items with two items.
- `fruits[1:3] = ["mango"]` replaces two items with one item, so the list gets shorter.
- `append` adds at the end. `insert(2, "dates")` puts an item at index `2` and shifts the later items right.
- `remove("mango")` deletes the first matching value. `pop()` deletes the last item. `pop(1)` deletes the item at index `1`.
- `clear()` deletes every item.
- `sort()` arranges the text alphabetically. `reverse()` flips that order.
- `count` tells how many times a value appears. `index` tells the first position of a value.


```python
fruits = ["apple", "banana", "cherry", "grapes"]
print(fruits)
fruits[1] = "mango"
print(fruits)
fruits[1:3] = ["orange", "pineapple"]
print(fruits)
fruits[1:3] = ["mango"]
print(fruits)
fruits.append("orange")
print(fruits)
fruits.insert(2, "dates")
print(fruits)
fruits.remove("mango")
print(fruits)
fruits.pop()
print(fruits)
fruits.pop(1)
print(fruits)
fruits.clear()
print(fruits)
fruits = ["pineapple", "apple", "banana", "cherry", "grapes"]
fruits.sort()
print(fruits)
fruits.reverse()
print(fruits)
print(fruits.count("apple"))
print(fruits.index("banana"))
```

## 4. Visit each item

`for fruit in fruits` gives you each value. `enumerate` gives you the position and the value together. The list here is the reversed list from the previous cell: `pineapple`, `grapes`, `cherry`, `banana`, `apple`.


```python
print(fruits)
for fruit in fruits:
    print(fruit)
for index, fruit in enumerate(fruits):
    print(index, fruit)
```

## 5. Build a list in a loop

Start with an empty list and `append` each new value. `i ** 2` is `i` squared, so this loop stores the squares of `0` through `9`.


```python
squares = []
for i in range(10):
    squares.append(i ** 2)
print(squares)
```

## 6. A comprehension builds the list in one line

`[i ** 2 for i in range(10)]` says: for each `i` in `0` through `9`, store `i` squared.

Add `if` to keep only some items. `[i for i in range(10) if i % 2 == 0]` keeps the even numbers.

Two `for` parts walk through two lists. For each drink, the inner loop pairs it with every size, so the menu has a small tea, a large tea, a small coffee, and a large coffee.

`len(word)` is the number of characters, so the last comprehension stores the length of each fruit.


```python
squares = [i ** 2 for i in range(10)]
print(squares)
```

```python
even_numbers = [i for i in range(10) if i % 2 == 0]
print(even_numbers)
```

```python
drinks = ["tea", "coffee"]
sizes = ["small", "large"]
menu = [f"{size} {drink}" for drink in drinks for size in sizes]
print(menu)
```

```python
fruits = ["apple", "banana", "cherry", "grapes"]
lengths = [len(word) for word in fruits]
print(lengths)
```
