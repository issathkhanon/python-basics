An ordered collection you treat as fixed.

This is lesson 6. A tuple keeps items in order, like a list. After you create it, you read items from it. You do not replace an item inside it. Parentheses `( )` create a tuple.

## Picture of the idea

A tuple keeps values in order, like a list. You can read a position. You cannot replace it.

![A tuple has positions you can read](diagrams/tuple_fixed.png)


## 1. An empty tuple

`tuple()` and `()` both create an empty tuple. `type` reports `tuple`.


```python
numbers = tuple()
print(type(numbers))
numbers = ()
print(type(numbers))
```

## 2. Positions and slices

The rules match a list. Counting starts at `0`, and a slice stops before the stop index.

`numbers` is `(2, 3, 4, 5)`.

- Index `1` is `3`.
- `1:` is `(3, 4, 5)`.
- `1:3` is `(3, 4)`.
- `:-2` drops the last two items, leaving `(2, 3)`.
- `::2` keeps every second item: `(2, 4)`.


```python
numbers = (2, 3, 4, 5)
print(numbers)
print(numbers[1])
print(numbers[1:])
print(numbers[1:3])
print(numbers[:-2])
print(numbers[::2])
```

## 3. Join and repeat

`+` joins two tuples into a new tuple. The originals stay as they are.

`*` repeats the items. `numbers_b * 2` writes the four numbers, then writes them again.


```python
numbers_b = (7, 8, 9, 10)
combined = numbers + numbers_b
print(combined)
```

```python
print(numbers_b * 2)
```

## 4. A tuple does not accept a new item in place

`numbers_b[2] = 20` asks Python to replace the `9`. A tuple refuses item assignment, so this cell raises `TypeError`. The tuple is still `(7, 8, 9, 10)` after the error.


```python
numbers_b[2] = 20
```

## 5. Count and find

`count(3)` tells how many times `3` appears. `index(4)` tells the first position of `4`, which is `2` because counting starts at `0`.


```python
print(numbers)
print(numbers.count(3))
print(numbers.index(4))
```

## 6. Pack values, then unpack them

Commas pack values into a tuple, even when you leave out the parentheses. `1, 3.14, "hello"` is one tuple with an integer, a float, and text.

Unpacking uses one name for each item, in order. `a, b, c = values` stores `1` in `a`, `3.14` in `b`, and `"hello"` in `c`.

A star gathers the leftover items into a list. `a, *rest = values` stores `1` in `a` and the remaining two items in `rest`.


```python
values = 1, 3.14, "hello"
print(values, type(values))
```

```python
first, second, third = values
print(first)
print(second)
print(third)
```

```python
first, *rest = values
print(first)
print(rest)
```

## 7. A tuple can contain tuples

`records[1]` is the whole middle tuple, `(4, 4, 6)`. `records[2][1]` goes into the third tuple and takes the item at index `1`, which is `False`.

The nested loops visit each inner tuple, then each item inside it.


```python
records = ((3, 3, 4), (4, 4, 6), (True, False))
print(records)
print(records[1])
print(records[2][1])
```

```python
for group in records:
    for item in group:
        print(item)
```
