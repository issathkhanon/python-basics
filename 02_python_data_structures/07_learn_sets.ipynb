# Sets

Unique items, with no position you can rely on.

This is lesson 7. A set stores each value once. It does not keep a position the way a list does, so you ask whether a value is present instead of asking for index `0`.

Run each code cell with **Shift+Enter**. One cell is written to fail.

Use the name `numbers` for your set. The name `set` belongs to the function that creates a set. Storing a dictionary in `set` makes a later `set()` call fail.

In this lesson you will:

- Create an empty set
- See a repeated value disappear
- Add and remove values
- Test membership
- Combine two sets with union, intersection, and difference


## 1. An empty set

`set()` creates an empty set. `{}` does not: empty braces create a dictionary. `type` reports `set` for the value below.


```python
numbers = set()
print(numbers)
print(type(numbers))
```

## 2. Repeated values are stored once

`set([10, 20, 30, 40, 50, 40])` reads six numbers and keeps `40` once. The print can show the numbers in any order. Order is not part of what a set promises.


```python
numbers = set([10, 20, 30, 40, 50, 40])
print(numbers)
```

## 3. Add one value

`add(60)` puts `60` into the set. Adding a value that is already present leaves the set unchanged, because a set keeps each value once.


```python
numbers.add(60)
print(numbers)
```

## 4. Remove a value that is present

`remove(20)` deletes `20`. The value has to be in the set.


```python
numbers.remove(20)
print(numbers)
```

## 5. `pop` removes one value

`pop()` removes a single value and returns it. You do not choose which one. The set printed underneath is what remains.


```python
print(numbers.pop())
print(numbers)
```

## 6. `remove` fails when the value is missing

`80` is not in the set. `remove(80)` raises `KeyError`. This cell is written to fail. Read the last line of the error.


```python
numbers.remove(80)
```

## 7. `discard` is quiet when the value is missing

`discard(80)` also means "delete `80` if it is there." The value is absent, so the set stays as it was and the cell does not raise an error.

`clear()` then deletes every remaining value. `set()` in the print means the set is empty.


```python
numbers.discard(80)
print(numbers)
```

```python
numbers.clear()
print(numbers)
```

## 8. Ask whether a value is present

`in` answers `True` or `False`. `30` is in `group_a`. `5` is not.


```python
group_a = {10, 20, 30, 40}
print(30 in group_a)
print(5 in group_a)
```

## 9. Combine two sets

`group_b` is `{15, 25, 35, 20, 40}`.

- `union` keeps every value from either set.
- `intersection` keeps values that are in both sets: `20` and `40`.
- `difference` keeps values from `group_a` that are absent from `group_b`: `10` and `30`.
- `issubset` asks whether every value of `group_a` is also in `group_b`. `10` and `30` are missing there, so the answer is `False`.

The printed order of a set can vary. Membership is the fact to read.


```python
group_b = {15, 25, 35, 20, 40}
print(group_a.union(group_b))
print(group_a.intersection(group_b))
print(group_a.difference(group_b))
print(group_a.issubset(group_b))
```
