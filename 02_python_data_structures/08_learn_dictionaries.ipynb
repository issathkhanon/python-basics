# Dictionaries

Values stored under keys.

This is lesson 8 of 8. A dictionary pairs each key with one value. You look up a value by its key, such as `"name"`, instead of by a position. Curly braces `{ }` with `key: value` pairs create a dictionary.

Run each code cell with **Shift+Enter**.

In this lesson you will:

- Create an empty dictionary
- See what happens when the same key is written twice
- Read, change, add, and delete pairs
- Loop over keys and values
- Share one dictionary between two names, and copy it
- Store a dictionary inside a dictionary


## 1. An empty dictionary

`{}` and `dict()` both create an empty dictionary. This is why an empty set needs `set()`: empty braces are already used for a dictionary.


```python
person = {}
print(type(person))
person = dict()
print(type(person))
```

## 2. One key keeps one value

`"name"` is written twice, first as `"khan"` and then as `"Issath"`. A dictionary keeps one value per key, and the later assignment replaces the earlier one. The print shows `"name"` paired with `"Issath"`. `"age"` and `"location"` appear once each.


```python
person = {"name": "khan", "age": 30, "location": "tamil nadu", "name": "Issath"}
print(person)
```

## 3. Read a value

`person["name"]` returns the value stored under that key. A missing key raises `KeyError`.

`get` also reads a value. When the key is missing, `get` returns `None`, or the fallback you provide. `"last_name"` is missing, so the second `get` returns `None` and the third returns `"Not Available"`.


```python
print(person["name"])
print(person.get("location"))
print(person.get("last_name"))
print(person.get("last_name", "Not Available"))
```

## 4. Change a value, add a key, and delete a key

Assigning an existing key replaces its value. Assigning a new key adds a pair. `del person["age"]` removes that pair.


```python
person["age"] = 33
person["location"] = "chennai"
person["state"] = "Tamil Nadu"
print(person)
```

```python
del person["age"]
print(person)
```

## 5. Loop over each pair

`items()` gives each key together with its value. After the deletion above, the pairs are `name`, `location`, and `state`.


```python
for key, value in person.items():
    print(f"{key} ---- {value}")
```

## 6. Two names can point at one dictionary

`shared = person` does not copy the pairs. Both names refer to the same dictionary. Changing `shared["name"]` to `"khan"` changes what `person` sees as well.


```python
shared = person
shared["name"] = "khan"
print(shared)
print(person)
```

## 7. `copy` makes a second dictionary

`person.copy()` creates a new dictionary with the same pairs. Changing `copied["name"]` to `"John"` leaves `person` holding `"khan"`.

This copy is shallow. The new dictionary has its own keys. If a value were itself a dictionary, both copies would still share that inner dictionary.


```python
copied = person.copy()
copied["name"] = "John"
print(copied)
print(person)
```

## 8. A dictionary inside a dictionary

Each person is a dictionary stored under a key. `people["person2"]` is Doe's dictionary. `people["person2"]["age"]` reads the `age` inside it, which is `35`.


```python
people = {
    "person1": {"name": "john", "age": 25},
    "person2": {"name": "Doe", "age": 35},
}
print(people)
print(people["person2"]["age"])
```

## 9. A dictionary comprehension

`{x: x ** 2 for x in range(10) if x % 2 == 0}` builds key and value together. Each even number from `0` through `8` becomes a key, and its square becomes the value.


```python
squares = {x: x ** 2 for x in range(10) if x % 2 == 0}
print(squares)
```
