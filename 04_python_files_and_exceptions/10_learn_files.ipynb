# Files

Read and write text on disk.

This is lesson 10. `open` connects your program to a file. The second argument is the mode: `r` reads, `w` replaces the file, `a` adds to the end, and `w+` replaces the file and also lets you read it.

`with` closes the file when the block ends, including when an error happens inside the block.

Run each code cell with **Shift+Enter**. These cells use `example.txt` in this project folder. Run them from the top, because each cell changes that file.

In this lesson you will:

- Write a file, then read it back
- Read one line at a time
- Replace a file and append a line
- Move back to the start of a file with `seek`


## 1. Write a file, then read it

`"w"` creates `example.txt` if it is missing, and replaces whatever is already in it. `write` stores the characters you give it, including `\n`, which starts a new line.

A second `with` opens the same file with `"r"` and `read()` returns the whole text as one string.


```python
with open("example.txt", "w") as file:
    file.write("Hello\n")
    file.write("How are you?\n")
    file.write("I am Issath Khan\n")

with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

## 2. Read one line at a time

`readlines()` returns a list of strings. Each string still ends with `\n`, so `print` would show a blank line between them. `strip()` removes that newline before printing.


```python
with open("example.txt", "r") as file:
    lines = file.readlines()
    for line in lines:
        print(line.strip())
```

## 3. `"w"` replaces the whole file

Opening with `"w"` again removes the three lines from the first cell. After this cell, `example.txt` contains only `Hello world`.


```python
with open("example.txt", "w") as file:
    file.write("Hello world\n")
```

## 4. `"a"` adds to the end

`"a"` keeps the text that is already there and writes after it. The file now has `Hello world` and then `How do you do?`.


```python
with open("example.txt", "a") as file:
    file.write("How do you do?\n")

with open("example.txt", "r") as file:
    print(file.read())
```

## 5. Write, then move back to the start

`"w+"` replaces the file and leaves it open for reading. After `write`, the position is at the end, so `read()` returns an empty string.

`seek(0)` moves that position back to the first character. The next `read()` returns both lines.


```python
with open("example.txt", "w+") as file:
    file.write("Everything is fine?\n")
    file.write("Hello world\n")
    print("Before seek:", repr(file.read()))
    file.seek(0)
    print(file.read())
```
