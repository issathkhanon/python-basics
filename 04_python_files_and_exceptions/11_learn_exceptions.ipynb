# Exceptions

What to do when a line fails.

This is lesson 11. An exception is Python stopping a line because it cannot finish it, such as dividing by zero or reading a missing file. `try` marks the lines that might fail. `except` runs when a matching exception happens.

These cells store the number directly, so you can read the result without typing. `input()` pauses for the keyboard and returns text. `int()` turns that text into a whole number, and the same `try` blocks handle the result.

Run each code cell with **Shift+Enter**.

In this lesson you will:

- Catch division by zero separately from other errors
- Run `else` only when the `try` block succeeds
- Use `finally` after either outcome
- See why `with` already closes a file before `finally` runs


## 1. Catch one kind of error

`1 / 0` raises `ZeroDivisionError`. That `except` prints `You entered zero`. The broader `except Exception` is skipped because the first matching `except` already handled the error.


```python
try:
    num = 0
    result = 1 / num
except ZeroDivisionError:
    print("You entered zero")
except Exception as ex:
    print(f"Some other exception: {ex}")
```

## 2. A different error uses the broader `except`

`int("khan")` cannot turn that text into a whole number, so it raises `ValueError`. `ValueError` is a kind of `Exception`, and it is not a `ZeroDivisionError`, so the second `except` prints the message.


```python
try:
    num = int("khan")
    result = 1 / num
except ZeroDivisionError:
    print("You entered zero")
except Exception as ex:
    print(f"Some other exception: {ex}")
```

## 3. `else` runs when nothing fails

`num` is `10`, so `1 / num` is `0.1`. Neither `except` runs. `else` on a `try` runs only after the `try` block finishes with no exception.


```python
try:
    num = 10
    result = 1 / num
except ZeroDivisionError:
    print("You entered zero")
except Exception as ex:
    print(f"Some other exception: {ex}")
else:
    print(f"The result: {result}")
```

## 4. `finally` runs after success or failure

`finally` runs after the `try` block, whether an exception happened or not.

`with` closes the file when its own block ends. By the time `finally` runs, `file.closed` is `True`. The test `not file.closed` is false, so `print("file closed")` is skipped.

`'file' in locals()` asks whether the name `file` was created. `locals()` is the dictionary of names in this cell. If `open` fails before `with` assigns `file`, that test is false and the cell does not touch a missing name.

`file.closed` is `True` or `False`. Writing `file.closed()` tries to call that boolean and raises `TypeError`.


```python
try:
    with open("example.txt", "r") as file:
        content = file.read()
except Exception as ex:
    print(f"Exception occurred: {ex}")
else:
    print(f"File content:\n{content}")
finally:
    if "file" in locals() and not file.closed:
        file.close()
        print("file closed")
    elif "file" in locals():
        print("Already closed by with:", file.closed)
```
