# Imports

Use code that already exists.

This is lesson 12. A module is a Python file. `import` loads that file and gives you its names. The standard library ships with Python. `numpy` is a separate package installed into the virtual environment.

Run each code cell with **Shift+Enter**. Run this notebook with its folder as the working directory.

In this lesson you will:

- Call a function through the module name
- Import a few names directly
- Give a module a shorter name
- Pick a random integer and a random item


## 1. Import a module and use its name

`import math` loads the math module. `math.sqrt(16)` calls the square-root function inside it. The module name stays in front, so you can see where `sqrt` came from.


```python
import math
math.sqrt(16)
```

## 2. Import the names you need

`from math import sqrt, pi` puts those two names into this cell. You then call `sqrt(25)` without writing `math.` first. `pi` is the constant 3.141592653589793.


```python
from math import sqrt, pi
print(sqrt(25))
print(pi)
```

## 3. Rename a module as you import it

`import numpy as np` loads NumPy and gives it the short name `np`. `np.array([2, 3, 4])` builds an array of those three numbers. This project installs NumPy from `requirements.txt`.


```python
import numpy as np
np.array([2, 3, 4])
```

## 4. `random` chooses for you

`random.randint(1, 7)` returns one integer from 1 through 7, including both ends. `random.choice` returns one item from the list. Run the cell again and the printed values can change.


```python
import random
print(random.randint(1, 7))
print(random.choice(["apple", "banana", "cherry"]))
```
