# Packages

Your own folder of modules.

This is lesson 13. A package is a folder of modules. This lesson's package is `mypackage`. Inside it, `maths.py` is a module, and `__init__.py` marks the folder as a package.

```text
mypackage/
    __init__.py
    maths.py
run_maths.py
```

`mypackage.maths` means the `maths` module inside the `mypackage` package.

Run each code cell with **Shift+Enter**, with this folder as the working directory.

Two scripts in this folder are meant for the terminal. From this folder, run:

```bash
python run_maths.py
python copy_text.py
```

`run_maths.py` prints `7` and `3`. `copy_text.py` copies `source.txt` to `destination.txt`.

## Picture of the idea

## 1. Import a function from your package

`addition` lives in `mypackage/maths.py`. `from mypackage.maths import addition` loads that function. `addition(4, 5)` returns `9`.

The same file also defines `subtraction`. The terminal script `run_maths.py` imports both names and prints their results.


````python
from mypackage.maths import addition, subtraction
print(addition(4, 5))
print(subtraction(5, 2))
````
