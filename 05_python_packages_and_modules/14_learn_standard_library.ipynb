## 1. `os` talks to the operating system

`os.getcwd()` returns the folder where this notebook is running. `os.mkdir("test_dir")` creates that folder and returns `None`, so the print shows `None`.

The last two lines check that the folder exists, then delete it, so the next run of this cell can create it again.


```python
import os
print(os.getcwd())
print(os.mkdir("test_dir"))
print(os.path.isdir("test_dir"))
os.rmdir("test_dir")
```

## 2. `shutil` copies a file

`shutil.copyfile` copies the bytes from `source.txt` into `destination.txt`. It returns the destination name. The terminal script `copy_text.py` does this same copy.


```python
import shutil
shutil.copyfile("source.txt", "destination.txt")
```

## 3. `json` turns a dictionary into text

`json.dumps` builds a string of JSON text. `type` reports `str`. `json.loads` reads that text and builds a dictionary again. `type` then reports `dict`.


```python
import json
data = {"name": "Issath", "age": 35}
json_str = json.dumps(data)
print(json_str)
print(type(json_str))
value = json.loads(json_str)
print(value)
print(type(value))
```

## 4. `csv` reads and writes rows

`csv.writer` writes each list as one comma-separated row. `newline=""` keeps the file from gaining an extra blank line on each row.

`csv.reader` gives each row back as a list of strings. The age is the text `"35"`, not the number `35`.


```python
import csv
with open("example.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "age"])
    writer.writerow(["Issath", "35"])

with open("example.csv", mode="r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## 5. `datetime` stores a moment

`datetime.now()` is the current date and time. The printed value changes each time you run the cell. `timedelta(days=1)` is a span of one day. Subtracting it produces the same clock time yesterday.


```python
from datetime import datetime, timedelta
now = datetime.now()
print(now)
print(type(now))
yesterday = now - timedelta(days=1)
print(yesterday)
```

## 6. `time` can pause

`time.time()` returns the number of seconds since 1 January 1970. `time.sleep(3)` waits 3 seconds. The second number is about 3 larger than the first.


```python
import time
print(time.time())
time.sleep(3)
print(time.time())
```

## 7. `re` finds a pattern in text

`r"\d+"` means one or more digits. `re.search` finds the first match in the sentence. `match.group()` returns the matched text, `123`.


```python
import re
pattern = r"\d+"
text = "There are 123 oranges"
match = re.search(pattern, text)
print(match.group())
```
