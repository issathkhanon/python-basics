# Classes

Objects that keep their own data.

This is lesson 15. A class is a blueprint. Each object made from it has its own attributes, such as a name and a color. A method is a function that belongs to the class and receives the object as `self`.

In this lesson you will:

- Store values on an object with `__init__`
- Read those values from methods
- See the difference between a method that prints and a method that returns

## Picture of the idea

A class is the blueprint. Each object made from it keeps its own values, and a method runs on that object.

![A class makes objects, and each object can run a method](diagrams/class_and_objects.png)


## 1. `__init__` stores the starting values

`Car("BMW", "Blue", 5)` creates an object and calls `__init__`. `self` is that object. `self.name = name` stores the value on it.

`get_name`, `get_color`, and `get_seats` return the stored values. `start_car` prints a message and does not use `return`.


```python
class Car:
    def __init__(self, name, color, seats):
        self.name = name
        self.color = color
        self.seats = seats

    def get_color(self):
        return self.color

    def get_name(self):
        return self.name

    def get_seats(self):
        return self.seats

    def start_car(self):
        print(f"{self.name} car engine starts")
```

## 2. Call a method that prints

`mycar1` and `mycar2` are two objects. Each one keeps its own name. `start_car` prints the message itself, so the call is not wrapped in `print`.

A method with no `return` hands back `None`. `print(mycar1.start_car())` would print the engine message and then print that `None`.


```python
mycar1 = Car("BMW", "Blue", 5)
mycar1.start_car()
mycar2 = Car("Benz", "Black", 7)
mycar2.start_car()
```

## 3. Call a method that returns a value

`get_name` returns text. `print` is what shows it. `get_seats` returns the number stored on that car.


```python
print(mycar1.get_name())
print(mycar1.get_color())
print(mycar2.get_seats())
```
