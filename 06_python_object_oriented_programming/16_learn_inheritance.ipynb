## 1. One parent: `super()`

`SelfDrivingCar` is a `Car` with one extra value, `is_self_driving`. `super().__init__(name, color, seats)` runs `Car.__init__` and stores the three car values. You do not pass `self`. `super()` finds the parent and passes the current object.

`Car.__init__(self, name, color, seats)` reaches that same parent method in this example. The child names `Car` and passes `self` itself.

`start_self_car` prints its own message when the car is self-driving. Otherwise `super().start_car()` runs the parent method.


````python
class Car:
    def __init__(self, name, color, seats):
        self.name = name
        self.color = color
        self.seats = seats

    def start_car(self):
        print(f"{self.name} car engine starts")


class SelfDrivingCar(Car):
    def __init__(self, name, color, seats, is_self_driving):
        super().__init__(name, color, seats)
        self.is_self_driving = is_self_driving

    def start_self_car(self):
        if self.is_self_driving:
            print(f"{self.name} supports self driving and engine starts")
        else:
            super().start_car()
````

````python
tesla = SelfDrivingCar("Tesla", "Black", 5, True)
tesla.start_self_car()
toyota = SelfDrivingCar("Toyota", "Black", 5, False)
toyota.start_self_car()
````

## 2. Two parents: call each class by name

`Dog` inherits from `Animal` and `Pet`. Each parent has its own `__init__`, and neither parent calls `super()`.

One `super()` call in `Dog` would follow the parent order and run only `Animal.__init__`. `owner` would never be stored. Naming each parent runs both initializers. You pass `self` because these are direct calls:

```python
Animal.__init__(self, name)
Pet.__init__(self, owner)
```


````python
class Animal:
    def __init__(self, name):
        self.name = name


class Pet:
    def __init__(self, owner):
        self.owner = owner


class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def sound(self):
        print(f"{self.name} says woof")
````

````python
dog = Dog("Pubby", "John")
print(dog.name)
print(dog.owner)
dog.sound()
````
