## 1. One function, two classes

`Rectangle.area` and `Square.area` both compute an area and return the number. `print_area` does not care which class the object came from. It calls `shape.area()` and prints the returned number.

`Shape.area` prints its own sentence and returns `None`. It is a separate class here. `print_area` is used with the rectangle and the square, whose `area` methods return numbers.


```python
class Shape:
    def area(self):
        print("The area of the shape")


class Rectangle:
    def __init__(self, width, length):
        self.width = width
        self.length = length

    def area(self):
        return self.width * self.length


class Square:
    def __init__(self, length):
        self.length = length

    def area(self):
        return self.length * self.length


def print_area(shape):
    print(f"area is {shape.area()}")


rectangle = Rectangle(5, 10)
print_area(rectangle)
square = Square(7)
print_area(square)
```

## 2. An abstract method must be implemented

`Vehicle` inherits from `ABC` and marks `start_engine` with `@abstractmethod`. Creating a `Vehicle` object fails because that method has no working body.

`Car` and `Bike` each define `start_engine`. `start_vehicle` calls that method on whichever object it receives.


```python
from abc import ABC, abstractmethod


class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass


Vehicle()
```

```python
class Car(Vehicle):
    def start_engine(self):
        print("Start the car engine")


class Bike(Vehicle):
    def start_engine(self):
        print("Start the bike engine")


def start_vehicle(vehicle):
    vehicle.start_engine()


start_vehicle(Car())
start_vehicle(Bike())
```

## 3. Protected: one leading underscore

Python has no `protected` keyword. One leading underscore is the protected access specifier.

`_age` is protected. `Person` may use it, and a child class such as `Employee` may use it. Code outside that class family should call the public method `get_age()`.

`get_name` and `get_age` have no leading underscore. They are public. Anyone with the object may call them.


```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self._age = age

    def get_name(self):
        return self.__name

    def get_age(self):
        return self._age


class Employee(Person):
    def show_age(self):
        print(self._age)
```

```python
employee = Employee("Khan", 30)
employee.show_age()
print(employee.get_age())
```

## 4. Private: two leading underscores

Python has no `private` keyword. Two leading underscores are the private access specifier.

`__name` is private to `Person`. Only a method of `Person`, such as `get_name`, may use it. Code outside the class cannot read `person.__name`. A child class cannot read `self.__name` either. The child still uses the protected `_age`, and it reads the private name through the public method.


```python
person = Person("Issath", 35)
print(person.__name)
```

```python
class Employee(Person):
    def show(self):
        print(self.__name)


employee = Employee("Khan", 30)
employee.show()
```

## Read a private attribute through a public method

`get_name` is a method of `Person`, so it is allowed to use the private `__name`. Outside the class, call that public method. Call `get_age()` for the age. The protected `_age` was already read inside `Employee.show_age`.


```python
print(employee.get_name())
print(employee.get_age())
```
