# Magic methods

Methods Python calls for you.

This is lesson 18. A magic method starts and ends with two underscores. You do not call `__str__` or `__add__` by those names. `print` calls `__str__`, `repr` calls `__repr__`, `+` calls `__add__`, and `==` calls `__eq__`.

## 1. The default text

A new class already has `__str__` and `__repr__` from `object`. Those methods return the class name and a memory address, so `print(car)` and `repr(car)` show the same line.

The names in `dir` that start and end with `__` are the magic methods already attached to the object.


```python
class Car:
    def __init__(self, name, color):
        self.name = name
        self.color = color

    def start_car(self):
        print(f"{self.name} car engine starts")
```

```python
car = Car("BMW", "RED")
print(car)
print(repr(car))
print([name for name in dir(car) if name in ("__str__", "__repr__", "__eq__")])
```

## 2. `__str__` for `print`, `__repr__` for `repr`

`print(car)` calls `__str__` and prints the sentence. `repr(car)` calls `__repr__` and prints a text that looks like the call used to build the car.

`__repr__` needs both underscores on each side. A method named `__rep__` is an ordinary method, so `print` never calls it and the address line comes back.


```python
class Car:
    def __init__(self, name, color):
        self.name = name
        self.color = color

    def start_car(self):
        print(f"{self.name} car engine starts")

    def __str__(self):
        return f"Name is {self.name} and Color is {self.color}"

    def __repr__(self):
        return f"Car(Name={self.name} Color={self.color})"
```

```python
car = Car("BMW", "RED")
print(car)
print(repr(car))
```

## 3. `__add__` runs for `+`

`v1 + v2` calls `v1.__add__(v2)` and returns a new `Vector`. This class has no `__str__`, so `print` uses `__repr__` and shows `Vector(6,8)`.

Without `__repr__`, that same addition still returns a vector, and `print` shows the memory address instead of the numbers.


```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vector({self.x},{self.y})"

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
```

```python
v1 = Vector(2, 3)
v2 = Vector(4, 5)
print(v1 + v2)
```

## 4. `__eq__` runs for `==`

`==` calls `__eq__`. The method compares `x` and `y` and returns `True` or `False`.

The name is `__eq__`, with two underscores on each side. `_eq__` is an ordinary method, so `==` would compare the objects' identities instead of their numbers.


```python
print(v1 + v2 == Vector(6, 8))
print(v1 == v2)
```
