# Custom exceptions

An exception class of your own.

This is lesson 19. `AgeException` inherits from `Exception`, so `raise` and `except` treat it like the built-in errors. `pass` means the class adds no extra behavior. The name `AgeException` is enough to tell this failure apart from `ValueError` or `ZeroDivisionError`.

The age is stored in the cell, so the notebook runs without waiting for the keyboard. `input()` returns whatever a person types, and `int()` turns that text into a number you can pass to `check_age`.

Run the cells from top to bottom with **Shift+Enter**.

In this lesson you will:

- Define an exception by inheriting from `Exception`
- Raise it when an age is outside 19 to 29
- Catch that exception by its own name


## 1. A class that is only an exception

`class AgeException(Exception)` creates a new error type. `check_age` prints a message when the age is greater than 18 and less than 30. Any other age runs `raise AgeException`.

`except AgeException` runs only for this error. A different exception would skip this `except`.


```python
class AgeException(Exception):
    pass


def check_age(age):
    if age > 18 and age < 30:
        print("You are eligible")
    else:
        raise AgeException
```

## 2. An age inside the range

`25` is greater than 18 and less than 30, so `check_age` prints the message. The `except` block does not run.


```python
try:
    check_age(25)
except AgeException:
    print("Enter an eligible age")
```

## 3. An age outside the range

`40` fails the test, so `check_age` raises `AgeException`. Python leaves the `try` block and runs the matching `except`, which prints `Enter an eligible age`.


```python
try:
    check_age(40)
except AgeException:
    print("Enter an eligible age")
```
