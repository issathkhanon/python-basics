# Iterators

Ask for one item, then the next.

This is lesson 20. A list keeps every song at once. An iterator is a bookmark in that list. `next` moves the bookmark forward and gives you one song.

Run the cells from top to bottom with **Shift+Enter**. Each example creates its own iterator, so an earlier cell cannot use up the songs in a later cell.

In this lesson you will:

- Turn a list into an iterator with `iter`
- Take one song at a time with `next`
- Notice when the songs are finished
- See that a `for` loop does those steps for you

## Picture of the idea

The list keeps every song. `iter` makes a bookmark. Each `next` moves the bookmark one song forward. After the last song, `next` raises `StopIteration`.

![next moves the bookmark from one song to the next](diagrams/song_bookmark.png)


## 1. A list shows every song

`print(songs)` shows the whole playlist. Nothing has been played yet. The list is still there after you print it.


```python
songs = ["Shape of You", "Believer", "Perfect"]
print(songs)
```

## 2. `next` plays one song

`iter(songs)` creates a bookmark at the start. The first `next` returns `Shape of You` and moves the bookmark. The second `next` returns `Believer`. The third returns `Perfect`.

The list `songs` is unchanged. Only the bookmark moved.


```python
playlist = iter(songs)
print(next(playlist))
print(next(playlist))
print(next(playlist))
print(songs)
```

## 3. There is no song after the last one

This cell uses a new playlist with two songs. The third `next` has nothing left, so Python raises `StopIteration`. The `except` prints a normal sentence instead of stopping the notebook.

`There are no element` was the same idea. The playlist is finished.


```python
playlist = iter(["Shape of You", "Believer"])
print(next(playlist))
print(next(playlist))
try:
    next(playlist)
except StopIteration:
    print("The playlist is finished")
```

## 4. A `for` loop calls `next` for you

`for song in songs` creates an iterator, calls `next` until `StopIteration`, and then stops. You see each song and you do not write `next` yourself.


```python
for song in songs:
    print(song)
```
