# Generators

A function that hands back one value and waits.

This is lesson 21. `yield` pauses the function and gives you one value. The next time you ask, the function continues from the line after `yield`. It does not build the whole result first.

Run the cells from top to bottom with **Shift+Enter**.

In this lesson you will:

- Write a function that yields one welcome message at a time
- Read those messages with `for` and with `next`
- See the function stop when the names are finished


## 1. `yield` pauses after one message

`welcome_messages` walks through the names. At `yield` it hands back one sentence and waits. Calling the function does not print the messages. It returns a generator, which is an iterator made by this function.

The argument is the list of names. Each name is used. A loop written as `range(3)` would ignore the names you passed in.


```python
def welcome_messages(names):
    for name in names:
        yield f"Welcome, {name}"


messages = welcome_messages(["Issath", "Khan"])
print(messages)
```

## 2. `for` asks for every message

The `for` loop calls `next` on the generator. The first pass continues the function until the first `yield` and prints `Welcome, Issath`. The second pass continues after that `yield` and prints `Welcome, Khan`. Then the function ends.


```python
for message in welcome_messages(["Issath", "Khan"]):
    print(message)
```

## 3. `next` shows the pause

`next(messages)` runs the function until `yield` and returns that one sentence. The second `next` returns the second sentence. The third `next` finds no more names, so Python raises `StopIteration` and the `except` prints `No more names`.


```python
messages = welcome_messages(["Issath", "Khan"])
print(next(messages))
print(next(messages))
try:
    next(messages)
except StopIteration:
    print("No more names")
```
