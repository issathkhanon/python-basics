# Decorators

Extra steps around a function you already have.

This is lesson 22. A decorator receives a function and returns a new function. The new function runs your extra steps, then calls the original function. The original function stays the place where the real work happens.

## 1. Return the wrapper, and call it later

`receipt` receives `sell`. `wrapper` prints the heading, calls `sell`, then prints `Thank you`. `return wrapper` hands that function back. It does not run the sale yet.

`return wrapper()` would run the sale immediately, while `receipt` is still setting up, and you would not get a function to call later.

`billed_sale = receipt(sell)` is the long form. `billed_sale("Tea", 20)` is the moment the receipt actually prints.


```python
def receipt(func):
    def wrapper(item, price):
        print("--- Receipt ---")
        func(item, price)
        print("Thank you")
    return wrapper


def sell(item, price):
    print(f"{item}: {price}")


billed_sale = receipt(sell)
billed_sale("Tea", 20)
```

## 2. `@receipt` writes that assignment for you

`@receipt` above `sell` means `sell = receipt(sell)`. The name `sell` now points at `wrapper`. Calling `sell("Coffee", 30)` prints the heading, the coffee line, and `Thank you`.


```python
@receipt
def sell(item, price):
    print(f"{item}: {price}")


sell("Coffee", 30)
```

## 3. A decorator can take its own argument

`@add_tax(0.1)` has three steps:

1. `add_tax(0.1)` receives the rate and returns `decorator`.
2. `decorator` receives `meal_price` and returns `wrapper`.
3. `wrapper` calls `meal_price`, then adds 10 percent.

`0.1` is saved in `rate`. A second function can use a different rate. `snack_price` uses `0.05`, so a price of `40` gains a tax of `2.0`.


```python
def add_tax(rate):
    def decorator(func):
        def wrapper(price):
            amount = func(price)
            tax = amount * rate
            print(f"Tax at {rate}: {tax}")
            return amount + tax
        return wrapper
    return decorator


@add_tax(0.1)
def meal_price(price):
    return price


@add_tax(0.05)
def snack_price(price):
    return price


print(meal_price(100))
print(snack_price(40))
```
