# Logging

A record of what the shop did.

This is lesson 23. `print` shows a message only while the notebook is open. `logging` writes the message into `app.log`, so you can read it later.

Run the cells from top to bottom with **Shift+Enter**. `basicConfig` takes effect once in a kernel. Restart the kernel before you run this lesson again, or the file will keep the messages from the first run.

In this lesson you will:

- Write five shop messages, from a small detail up to a reason to close
- Read `app.log` back so the notebook shows what was saved
- Keep a quieter logger that stores only warning and above

## Picture of the idea

Each level is more serious than the one above it. A logger keeps that level and every level below it.

![Logging levels from DEBUG down to CRITICAL](diagrams/log_levels.png)


## 1. Five messages, then read the file

`level=logging.DEBUG` keeps every message. `filename="app.log"` sends them to that file instead of the notebook. `filemode="w"` starts the file over.

The five calls use five different levels, and each message says something different. The `with` block reads the file and prints it, which is why you see the log below the cell.

`tea-shop` is a second logger. Its level is `WARNING`, so the cup count and the biscuit sale are left out. The biscuit-tin warning is serious enough, so it is added at the end of the file.


```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    filename="app.log",
    filemode="w",
    format="%(levelname)s: %(message)s",
)

logging.debug("Checking that the tea kettle is hot")
logging.info("Sold one cup of tea")
logging.warning("Milk is running low")
logging.error("The card payment failed")
logging.critical("The shop has to close")

shop = logging.getLogger("tea-shop")
shop.setLevel(logging.WARNING)
shop.debug("Counting the cups still on the shelf")
shop.info("Sold one biscuit")
shop.warning("The biscuit tin is nearly empty")

with open("app.log") as file:
    print(file.read(), end="")
```
