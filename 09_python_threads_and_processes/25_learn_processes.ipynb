# Processes

The same kind of work, in separate programs.

This is lesson 25. A process runs a function in its own program. On this Mac, Python starts that program by importing the file again. The import must not start more processes, so `start` and `join` stay inside `if __name__ == "__main__"`. The functions stay outside that block, because the new process needs to import them.

![Processes start only when you run the file, then join waits](diagrams/processes.png)

The example is in [25_learn_processes.py](25_learn_processes.py). From this folder, run:

```bash
python 25_learn_processes.py
```
