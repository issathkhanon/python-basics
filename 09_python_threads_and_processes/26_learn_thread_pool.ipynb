# Thread pool

Many tasks, and only a few workers at a time.

This is lesson 26. A pool keeps a fixed number of threads. `map` gives the function one value from a sequence and returns the results in that same order. `submit` is one call. `result` waits for the value that call returns.

![map runs many tasks on a few workers, and submit runs one call](diagrams/thread_pool.png)

`with` closes the pool when the block ends. It waits for the workers and then stops them.

The example is in [26_learn_thread_pool.py](26_learn_thread_pool.py). From this folder, run:

```bash
python 26_learn_thread_pool.py
```
