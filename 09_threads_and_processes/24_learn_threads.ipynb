# Threads

Two workers inside one program.

This is lesson 24. A thread runs a function while the rest of the program keeps going. `start` begins that work. `join` waits until the worker has finished.

![Two workers start together, and join waits for both](diagrams/threads.png)

Both workers share this one program, so this lesson does not need `if __name__ == "__main__"`.

The example is in [24_learn_threads.py](24_learn_threads.py). From this folder, run:

```bash
python 24_learn_threads.py
```
