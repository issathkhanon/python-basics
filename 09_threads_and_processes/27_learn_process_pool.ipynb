# Process pool

Many tasks, each one in its own program.

This is lesson 27. `ProcessPoolExecutor` gives each task to a process. `map` still returns results in the order of the tasks. Creating the pool stays inside `if __name__ == "__main__"`, for the same reason as a single process: the new program imports this file and must not start another pool.

![A process pool starts inside __main__, maps the tasks, then closes](diagrams/process_pool.png)

`with` closes the pool at the end of the block.

The example is in [27_learn_process_pool.py](27_learn_process_pool.py). From this folder, run:

```bash
python 27_learn_process_pool.py
```
