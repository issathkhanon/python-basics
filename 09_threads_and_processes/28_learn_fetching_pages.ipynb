# One value per call

`map` hands the function one complete value each time.

This is lesson 28. Each call receives one whole item, such as one URL. The work spends its time waiting for a website, so threads are a good fit. `with` closes the pool after the calls return.

![map passes one whole value into each call](diagrams/one_value.png)

The example is in [28_learn_fetching_pages.py](28_learn_fetching_pages.py). From this folder, run:

```bash
python 28_learn_fetching_pages.py
```
