# Flask

A class register on your computer.

This is lesson 30. Flask connects an address to a function. A browser address returns an HTML page. A `/students` address returns the class list as JSON.

![A request matches a path and a method, then the function returns a page or JSON](diagrams/route.png)

The example is in [30_learn_flask.py](30_learn_flask.py). From this folder, run:

```bash
python 30_learn_flask.py
```

Open http://127.0.0.1:5000/

The list starts with Issath. Stopping the script and running it again brings that first list back. The names added while the script is running stay in memory until then.


## 1. One function for one address

`@app.route("/")` runs `home` when the address is `/`. `render_template("home.html")` reads `templates/home.html`. The script sets `template_folder` to the `templates` folder beside this file.

`if __name__ == "__main__"` starts the server only when you run this file.


## 2. The form sends a name

`/register` accepts GET and POST.

- GET returns the form in `templates/register.html`.
- POST reads `request.form["name"]`.

The input on that page is `name="name"`. That word is the key Flask stores in `request.form`. The function adds the student to the same list and returns `templates/added.html`. `{{ name }}` in that page is the name that was typed, so the page reads `Khan is on the register.`


## 3. The same list as JSON

`jsonify` turns the Python list into a JSON response. These commands talk to the running script.

```bash
curl http://127.0.0.1:5000/students
```

```bash
curl http://127.0.0.1:5000/students/1
```

```bash
curl -X POST http://127.0.0.1:5000/students \
  -H "Content-Type: application/json" \
  -d '{"name": "Khan"}'
```

```bash
curl -X PUT http://127.0.0.1:5000/students/1 \
  -H "Content-Type: application/json" \
  -d '{"name": "Khan"}'
```

```bash
curl -X DELETE http://127.0.0.1:5000/students/1
```

POST reads `request.json`. That is the JSON body, `{"name": "Khan"}`. The form used `request.form` because the browser sends form fields.

`students.append` adds a dictionary to the list the name already points at.


## 4. The address name matches the parameter

`/students/<int:student_id>` takes the number from the address and passes it into `get_student(student_id)`. The word in the route and the parameter are the same word. Flask calls the function with that name.

A missing id answers with `{"error": "student not found"}` and status `404`.


## 5. Delete builds a new list

`delete_student` does `students = ...`. That assignment builds a new list, so the function starts with `global students`. The new list replaces the one at the top of the file.

`get_students`, `add_student`, and `update_student` read the list or call `append`. Those steps use the list the name already points at, so those functions do not use `global`.


## Try it yourself

With the script running:

1. Open http://127.0.0.1:5000/register and add your name. Then `GET /students`.
2. `PUT` a new name onto student `1`.
3. `DELETE` that student and `GET /students` again.
