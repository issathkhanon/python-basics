# pytest

A mark book, checked by a separate test file.

This is lesson 31. [marks.py](marks.py) holds the class `MarkBook`. `add` stores a mark, `total` adds the stored marks, and `passed` is `True` when the average is `50` or more.

[tests/test_marks.py](tests/test_marks.py) builds the object in a fixture named `book`. pytest runs that fixture before each test and passes the new `MarkBook("Issath")` into the test. Issath with `40` and `60` has a total of `100` and passes. Issath with `20`, `30`, and `50` also totals `100`, and the average is below `50`, so `passed` is `False`. A mark of `75`, `50`, or `100` passes. A mark of `49` does not. `add(150)` and `add(-1)` raise `ValueError`. `pytest.raises` passes only when that error is raised, and the mark is not stored.

From this folder, run:

```bash
pytest
```

pytest finds `tests/test_marks.py` because the file name starts with `test_`. The test file adds the `12_testing` folder to the path so `marks.py` can be imported. Each passing test prints a dot. When an `assert` is false, pytest stops that test and prints the two values.


## Try it yourself

1. In `marks.py`, add a method `highest` that returns the largest stored mark. An empty book returns `0`.
2. In `tests/test_marks.py`, use the `book` fixture. Add `40` and `60`, then check that `highest()` is `60`.
3. Run `pytest` again and read the dots.
