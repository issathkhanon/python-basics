# uv

The tool that runs this project's commands.

This is lesson 32. uv installs the libraries named in `pyproject.toml` and runs a command inside the project environment. You do not activate a virtual environment first.


## 1. Install uv

On this Mac the install command is:

```bash
brew install uv
```

After that, `uv` is a command in the terminal.


## 2. The files uv keeps for this project

Three files at the project root belong to uv:

- `pyproject.toml` names the project and lists the libraries, including `ruff` and `pytest`.
- `uv.lock` records the exact version of each library.
- `.python-version` says `3.12`, the Python version for this project.

uv creates a folder named `.venv` and installs those libraries there.


## 3. `uv add` writes a library into the project

Run these commands from the project root, the folder that contains `pyproject.toml`.

One library:

```bash
uv add ruff
```

uv adds `ruff` to the `dependencies` list in `pyproject.toml`, records the chosen version in `uv.lock`, and installs it into `.venv`.

Every library named in a requirements file:

```bash
uv add -r requirements.txt
```

`-r` means read this file. Each line, such as `flask==3.1.3` and `pytest==9.2.0`, is added the same way as a single `uv add`. This project's `requirements.txt` was written by `uv export`, so those same versions are already listed in `pyproject.toml`.

Remove one library with the opposite command:

```bash
uv remove ruff
```

uv deletes that name from `pyproject.toml`, updates `uv.lock`, and removes the library from `.venv`.


## 4. `uv lock` and `uv sync`

`uv lock` reads `pyproject.toml` and writes `uv.lock`. The lock file is the exact version of every library, including the libraries those libraries need. It does not install anything.

```bash
uv lock
```

`uv sync` installs whatever `uv.lock` lists into `.venv`. Use it on a new checkout, after a pull, so the environment matches the lock file.

```bash
uv sync
```

`uv add` and `uv remove` already lock and sync. You run `uv lock` or `uv sync` on their own when the files and the environment have drifted apart.


## 5. `uv export` writes `requirements.txt`

This is the command recorded on the first line of `requirements.txt`:

```bash
uv export --no-dev --no-hashes --no-annotate --no-emit-project -o requirements.txt
```

It reads `uv.lock` and writes one library per line. `--no-dev` leaves out development libraries. `--no-hashes` leaves out the long hash lines. `--no-annotate` leaves out the comment after each name. `--no-emit-project` leaves out this project itself. `-o requirements.txt` is the file to write.

The other direction is `uv add -r requirements.txt`: the file's lines go back into `pyproject.toml`.


## 6. `uv venv` and `uv pip install`

These two commands are in the shell history. They fill an environment. They do not edit `pyproject.toml`.

```bash
uv venv
uv venv --python 3.12
```

The first creates `.venv` with the default Python. The second creates it with Python 3.12, the version in `.python-version`.

```bash
uv pip install -r requirements.txt
```

This installs the lines of `requirements.txt` into the environment. `uv add -r requirements.txt` does that install and also writes the names into `pyproject.toml`.


## 7. `uv run` uses the project environment

From the `13_python_uv_and_ruff` folder:

```bash
uv run ruff check 33_learn_before_ruff.py
```

`uv run` looks up `ruff` in the project `.venv` and runs it on that file. The next lesson reads the report.

If another environment is already active, uv ignores it and prints a warning like this:

```text
warning: `VIRTUAL_ENV=.../.uv_venv` does not match the project environment path `.../.venv` and will be ignored; use `--active` to target the active environment instead
```

The command still runs. It uses `.venv`, the environment that matches `pyproject.toml`.
