# Ruff

A checker for a Python file.

This is lesson 33. Ruff reads the file and prints a code for each problem. It does not run the program. [33_learn_before_ruff.py](33_learn_before_ruff.py) is the file with the problems. [33_learn_after_ruff.py](33_learn_after_ruff.py) is the same idea after the safe fixes and the edits you make by hand.

From this folder:

```bash
uv run ruff check 33_learn_before_ruff.py
```


## 1. A name that is never used

An earlier file had a function that stored `x` and `y` and then only printed a greeting. Ruff printed `F841` for each name:

```text
F841 Local variable `x` is assigned to but never used
 --> 33_learn_ruff.py:3:5
  |
2 | def hello():
3 |     x = 10
  |     ^

F841 Local variable `y` is assigned to but never used
 --> 33_learn_ruff.py:4:5
```

The caret points at the name. `F841` means the function stores a value and never reads it. Removing `x` and `y` made that check print `All checks passed!`.


## 2. The score file before the fixes

`33_learn_before_ruff.py` writes random scores. This is the report from `uv run ruff check` on that file. Seven problems, and three of them have a star, which means `--fix` can edit them.

```text
I001 [*] Import block is un-sorted or un-formatted
 --> 33_learn_before_ruff.py:1:1
  |
1 | / import random, json
2 | | import sys
help: Organize imports
  - import random, json
1 + import json
2 + import random
3 | import sys

F401 [*] `json` imported but unused
F401 [*] `sys` imported but unused

SIM115 Use a context manager for opening files
 --> 33_learn_before_ruff.py:6:12
6 |     file = open(filename, "w")

SIM102 Use a single `if` statement instead of nested `if` statements
12 |             if score > 50:
13 |                 if score % 2 == 0:
help: Combine `if` statements using `and`

TRY002 Create your own exception
18 |             raise Exception("No random scores were generated...")

BLE001 Do not catch blind exception: `Exception`
21 |     except Exception as e:

Found 7 errors.
[*] 3 fixable with `--fix`.
```

Read each code as one lesson:

- `I001` wants one import on each line, alphabetical order, and two blank lines before the function.
- `F401` means `json` and `sys` are imported and never used.
- `SIM115` wants `with open(...)` so the file closes when the block ends.
- `SIM102` wants `if score > 50 and score % 2 == 0`.
- `TRY002` wants your own error class instead of `raise Exception`.
- `BLE001` wants `except FileNotFoundError`, the error this function can actually meet. `except Exception` also catches every other error.


## 3. `--fix` edits the safe lines

```bash
uv run ruff check --fix 33_learn_before_ruff.py
```

The run on the score file ended with:

```text
Found 7 errors (3 fixed, 4 remaining).
1 unsafe fix available with `--fix --unsafe-fixes`.
```

The three fixed lines are the imports: unused `json` and `sys` are removed, and `import random` is left on its own line. The other four stay. `SIM102` is marked unsafe because joining the two `if` tests can change which lines run. You edit those four in the file.


## 4. The same idea after the edits

`33_learn_after_ruff.py` is that file once the remaining notes are applied:

- `class EmptyScoreError(ValueError)` replaces `raise Exception`.
- `with open(filename, "w")` replaces `open` and `file.close()`.
- one `if` uses `and`.
- `except FileNotFoundError` replaces `except Exception`.

One check can still remain. With a single blank line between `import random` and the class, Ruff prints:

```text
I001 [*] Import block is un-sorted or un-formatted
 --> 33_learn_after_ruff.py:1:1
1 | import random
2 |
3 | class EmptyScoreError(ValueError):
help: Organize imports
2 |
3 +
4 | class EmptyScoreError(ValueError):
```

The `+` is one more blank line. A top-level class needs two blank lines after the imports.


## 5. `ruff format` fixes spacing

```bash
uv run ruff format 33_learn_after_ruff.py
uv run ruff check 33_learn_after_ruff.py
```

The format command prints:

```text
1 file reformatted
```

The check then prints:

```text
All checks passed!
```

Run format again on a file that is already tidy and the message is `1 file left unchanged`.


## Try it yourself

1. In `33_learn_before_ruff.py`, add `import sys` on its own line and run `uv run ruff check` on that file. Read the `F401` line.
2. Put the two `if` tests back, one inside the other, in `33_learn_after_ruff.py`. Run the check and find `SIM102`.
3. Run `uv run ruff format 33_learn_after_ruff.py` and read whether the file was reformatted or left unchanged.
